```python
import numpy as np
import pandas as pd
```

# Preparing the Data

```python
df = pd.read_csv("ai_ds_job_salaries_2026_dirty.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum
```

```python
df.columns = df.columns.str.strip()
```

```python
df = df.replace(r'^\s*$', np.nan, regex=True)
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(keep=False)]
```

# Filling Null values in years_experience Column

```python
df['years_experience'].fillna(df['years_experience'].mean(), inplace=True)
```

```python
minus = df['years_experience'] < 0
minus.sum()
```

```python
df.loc[df["years_experience"] < 0, "years_experience"] = 0
```

```python
minus = df['years_experience'] < 0
minus.sum()
```

```python
df["years_experience"] = df["years_experience"].round().astype("int")
```

# Filling Null values in education_level Column

```python
df['education_level'].isnull().sum()
```

```python
df['education_level'] = df['education_level'].fillna('Unknown')
```

```python
df['education_level'].isnull().sum()
```

# Filling Null values in primary_language Column

```python
df['primary_language'].isnull().sum()
```

```python
df['primary_language'] = df['primary_language'].fillna('Python')
```

```python
df['primary_language'].isnull().sum()
```

# Converting data type

```python
df["ai_tools_hours_per_week"] = df["ai_tools_hours_per_week"].round().astype("int")
```

# Invalid Team Size

```python
minus = df['team_size'] < 0
minus.sum()
```

```python
minus1 = df['team_size'] > 100
minus1.sum()
```

```python
df['team_size'].median()
```

```python
df.loc[df["team_size"] < 0, "team_size"] = df['team_size'].median()
df.loc[df["team_size"] > 100, "team_size"] = df['team_size'].median()
```

```python
minus = df['team_size'] < 0
minus.sum()
```

```python
minus1 = df['team_size'] > 100
minus1.sum()
```

# Invalid Certifications

```python
df.loc[(df["certifications_count"] < 0) | (df["certifications_count"] > 10), "certifications_count"] = np.nan
```

```python
df["certifications_count"] = df["certifications_count"].fillna(df["certifications_count"].median())
```

```python
df["certifications_count"] = df["certifications_count"].round().astype("int")
```

# Remove Duplicates

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
```

```python
df.shape
```

# Categorical Values

```python
categorical_cols = df.select_dtypes(include="object").columns

with pd.option_context("display.max_rows", None):
    for col in categorical_cols:
        print(f"\n--- {col} ---")
        display(df[col].value_counts())
```

# Adding Country Name

```python
country_map = {
    "US": "United States",
    "IN": "India",
    "GB": "Great Britain",
    "DE": "Germany",
    "CA": "Canada",
    "AU": "Australia",
    "BR": "Brazil",
    "NL": "Netherlands",
    "FR": "France",
    "PL": "Poland",
    "SG": "Singapore",
    "ES": "Spain",
}

df["company_location_name"] = df["company_location"].map(country_map)
```

```python
col_to_move = df.pop("company_location_name")
target_idx = df.columns.get_loc("company_location")
df.insert(target_idx + 1, "company_location_name", col_to_move)
```

# Reset Index

```python
df = df.reset_index(drop=True)
```
